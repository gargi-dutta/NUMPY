## ADVANCE OPERATION WITH BUSINESS EXAMPLE

```python
import numpy as np
import matplotlib.pyplot as plt 
#👉 Matplotlib = Python library used to create graphs and visualize data.
#In short: np → arrays & calculations | plt → graphs & visualization.
#matplotlib → the main plotting library
#pyplott → module that provides plotting functions
#plt → short nickname (alias) for pyplot
```

```python
# Data structure : [resturant_id, 2021, 2022, 2023, 2024]
sales_data=np.array([
    [1, 150000, 180000, 220000, 250000], # chai hub
    [2, 120000, 140000, 260000, 190000], #biriyani
    [3, 160000, 190000, 240000, 240000], #pizza
    [4, 140000, 170000, 290000, 250000], #chicken
    [5, 100000, 160000, 260000, 220000]  #mutton
])

print("=== Zomato sales analysis ===")
print("\n sales data shape",sales_data.shape)
print("\n SAMPLE DATA FOR 1ST 3 RESTAURANTS: ",sales_data[:,1:]) # [:,1:] meaning : all select
# 1: meaning indexing 1 theke suru hobe
print("\n SAMPLE DATA FOR 1ST 3 RESTAURANTS: ",sales_data[:3])
```

TOTAL SALES PER YEAR

```python
print(np.sum(sales_data[:,1:],axis=0))
year_total=np.sum(sales_data[:,1:],axis=0)
print("\n YEARLY TOTAL: ",year_total)
all_total=np.sum(year_total,axis=0)
print("\n ALL TOTAL: ",all_total)
```

MINIMUM SALES PER RESTURANT

```python
min_sal = np.min(sales_data[:, 1:], axis=1) #left to right
print("\n MIN SALES: ",min_sal)


#MAXIMUM SALES PER YEAR

max_sales = np.max(sales_data[:,1:],axis=0) # top to bottom
print("\n max sales: ",max_sales)

# AVARAGE SALES PER RESTURENT

avg_sal=np.mean(sales_data[:,1:],axis=1)
print("\n AVARAGE SALE: ",avg_sal)
```

```python
cumulative_sum = np.cumsum(sales_data[:,1:],axis=1)
print("\n CUMULATIVE sales: ", cumulative_sum)

plt.figure(figsize=(8,6))
plt.plot(np.mean(cumulative_sum,axis=0))
plt.title("AVG CUMU SALES ACCROSS ALL RESTURANT")
plt.xlabel("YEARS")
plt.ylabel("SALES")
plt.grid(True)
plt.show()
```

```python
vector1 = np.array([1,2,3,4,5])
vector2 = np.array([6,7,8,9,10])

print("VECTOR ADDITION: ",vector1+vector2)

print("\n VECTOR MULTIPLICATION: ",vector1*vector2)

print("\n DOT PRODUCT ", np.dot(vector1, vector2))

angle = np.arccos(np.dot(vector1,vector2)/(np.linalg.norm(vector1)*np.linalg.norm(vector2)))
print("\n ANGLE: ",angle)
```
